```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

```python
classification_metrics = pd.read_csv(
    "../results/classification_metrics.csv"
)

classification_metrics
```

```python
best_model = classification_metrics.sort_values(
    by="Accuracy",
    ascending=False
)

best_model
```

## Classification Model Comparison

The performance of the classification models was compared using multiple evaluation metrics.

The comparison enables identification of the best-performing model based on Accuracy, Precision, Recall, F1-Score, and ROC-AUC.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=classification_metrics,
    x="Model",
    y="Accuracy"
)

plt.title("Classification Model Accuracy Comparison")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=classification_metrics,
    x="Model",
    y="Precision"
)

plt.title("Precision Comparison")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=classification_metrics,
    x="Model",
    y="Recall"
)

plt.title("Recall Comparison")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=classification_metrics,
    x="Model",
    y="F1 Score"
)

plt.title("F1 Score Comparison")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=classification_metrics,
    x="Model",
    y="ROC-AUC"
)

plt.title("ROC-AUC Comparison")

plt.show()
```

```python
best = classification_metrics.sort_values(
    by="Accuracy",
    ascending=False
).iloc[0]

print("===== Best Classification Model =====")

print("Model :", best["Model"])
print("Accuracy :", best["Accuracy"])
print("Precision :", best["Precision"])
print("Recall :", best["Recall"])
print("F1 Score :", best["F1 Score"])
print("ROC-AUC :", best["ROC-AUC"])
```

# Conclusion

The developed classification models were successfully evaluated using multiple performance metrics.

Among the evaluated algorithms, the best-performing model was selected based on overall predictive performance. The comparison demonstrates the effectiveness of evaluating multiple models before selecting a final machine learning solution.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load classification metrics
classification_metrics = pd.read_csv(
    "../results/classification_metrics.csv"
)

# Display results
display(classification_metrics)

# Select metrics for comparison
metrics = ["Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"]

# Plot
ax = classification_metrics.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(12, 7)
)

plt.title("Classification Model Performance Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(title="Evaluation Metrics")
plt.tight_layout()
plt.show()
```
